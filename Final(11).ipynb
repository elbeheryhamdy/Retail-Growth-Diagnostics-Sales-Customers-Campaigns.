```python
import csv
from io import StringIO

def psql_insert_copy(table, conn, keys, data_iter):
    with conn.connection.cursor() as cur:
        buf = StringIO()
        csv.writer(buf).writerows(data_iter)
        buf.seek(0)
        cols = ", ".join('"{}"'.format(k) for k in keys)
        cur.copy_expert(sql=f'COPY "{table.name}" ({cols}) FROM STDIN WITH CSV', file=buf)

tables = {
    "campaign_desc": campaign_desc, "campaign_table": campaign_table,
    "coupon_redempt": coupon_redempt, "coupon": coupon, "product": product,
    "hh_demographic": hh_demographic, "transaction_data": transaction_data,
    "causal_data": causal_data,
}

for name, df in tables.items():
    df.rename(columns=str.lower).to_sql(
        name, engine, if_exists="replace", index=False,
        chunksize=200_000, method=psql_insert_copy)
    print(f"✅ {name}: {len(df):,} صف")
```
